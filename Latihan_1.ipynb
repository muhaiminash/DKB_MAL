Bina program untuk meminta nama dan umur pengguna.

```python
nama = input("Masukkan nama anda: ")
umur = int(input("Masukkan umur anda: "))

print("Nama anda ialah", nama)
print("Umur anda ialah", umur, "tahun")
```

Bina program untuk mengira luas segi empat.

```python
panjang = float(input("Masukkan panjang: "))
lebar = float(input("Masukkan lebar: "))

luas = panjang * lebar

print("Luas segi empat ialah", luas)
```

Bina program untuk menentukan nombor ganjil atau genap.

```python
nombor = int(input("Masukkan satu nombor: "))

if nombor % 2 == 0:
    print(nombor, "ialah nombor genap.")
else:
    print(nombor, "ialah nombor ganjil.")
```

Paparkan nombor 1 hingga 10 menggunakan gelung.

```python
for nombor in range(1, 11):
    print(nombor)
```

Bina program untuk menentukan status lulus atau gagal berdasarkan markah.

```python
markah = float(input("Masukkan markah anda: "))

if markah >= 50:
    print("Status: Lulus")
else:
    print("Status: Gagal")
```
