#Bahagian B: Amali Numpy
Bina Array Markah


```python
import numpy as np

markah = np.array([70,85,90,65,80])
print(markah)
```

Kira Purata Markah



```python
purata = np.mean(markah)
print("Markah purata:", purata)
```

Cari Markah tertinggi

```python
tertinggi = np.max(markah)
print("Markah tertinggi:", tertinggi)
```

Markah Terendah

```python
terendah = np.min(markah)
print("Markah terendah:", terendah)
```

Kira Jumlah Markah keseluruhan

```python
jumlah = np.sum(markah)
print("Jumlah Markah Keseluruhan:", jumlah)
```

Tambah markah bonus

```python
markah_baru = markah + 5
print(markah_baru)
```

#BAHAGIAN C: Amali Pandas

Sediakan data pelajar



```python
import pandas as pd

data = {
    "Nama": ["Akmal","Adam","Nabil","Danish","Ameer"],
    "Markah": [70,85,90,65,80]
    }
```

Bina Data Frame

```python
df = pd.DataFrame(data)

print(df)
```

Paparkan Data teratas, nama pelajar dan markah

```python
print(df.head())
print("ini adalah untuk data teratas")
print("")
print("")
print(df["Nama"])
print("memaparkan nama sahaja")
print("")
print("")
print(df["Markah"])
print("memaparkan markah sahaja tanpa nama")
```

Kira purata markah menggunakan pandas

```python
purata = df["Markah"].mean()

print("Purata Markah:", purata)
```

Tambah Lajur Status

```python
df["Status"] = np.where(df["Markah"] >= 80, "Lulus", "Gagal")

print(df)
```

Tambah Laju Gred

```python
def tentukan_gred(markah):
  if markah >= 80:
    return "A"
  elif markah >= 65:
    return "B"
  elif markah >= 50:
    return "C"
  else:
    return "D"

df["Gred"] = df["Markah"].apply (tentukan_gred)

print(df)
```
