#Semua bar menggunakan satu warna

```python
import matplotlib.pyplot as plt

nama = ["Ali","Siti","Kumar","Meli Ling","Aiman"]
markah = [70,80,90,65,80]

plt.bar(nama, markah, color="blue")

plt.title("Markah Pelajar")
plt.xlabel("Nama Pelajar")
plt.ylabel("Markah")

plt.show()

#Code (color="blue")bole ditukar kepada=
#color="red"
#color="green"
#color="orange"
#color="purple"
#color="pink"
#color="yellow"
```

#Setiap bar menggunakan warna berbeza

```python
import matplotlib.pyplot as plt

nama = ["Ali", "Siti", "Kumar", "Mei Ling", "Aiman"]
markah = [70, 85, 90, 65, 80]

warna = ["blue", "green", "orange", "purple", "red"]
plt.bar(nama, markah, color=warna)

plt.title("Markah Pelajar")
plt.xlabel("Nama Pelajar")
plt.ylabel("Markah")

plt.show()
```

#Pengguna memilih warna graf

```python
import matplotlib.pyplot as plt

nama = ["Ali", "Siti", "Kumar", "Mei Ling", "Aiman"]
markah = [70, 85, 90, 65, 80]

warna_pilihan = input(
"Masukkan warna graf seperti blue, red, green atau orange: "
)

plt.bar(nama, markah, color=warna_pilihan)

plt.title("Markah Pelajar")
plt.xlabel("Nama Pelajar")
plt.ylabel("Markah")

plt.show()
```

#Ubah warna bar, bingkai dan latar belakang

```python
import matplotlib.pyplot as plt

nama = ["Ali", "Siti", "Kumar", "Mei Ling", "Aiman"]
markah = [70, 85, 90, 65, 80]

plt.bar(
nama,
markah,
color="skyblue",
edgecolor="black",
linewidth=2
)

plt.title("Markah Pelajar")
plt.xlabel("Nama Pelajar")
plt.ylabel("Markah")

plt.gca().set_facecolor("lightyellow")

plt.show()
```

#Ubah warna berdasarkan markah

```python
import matplotlib.pyplot as plt

nama = ["Ali", "Siti", "Kumar", "Mei Ling", "Aiman"]
markah = [70, 85, 90, 65, 80]

warna = []

for nilai in markah:
  if nilai >= 80:
    warna.append("green")
  else:
    warna.append("red")

plt.bar(nama, markah, color=warna)

plt.title("Markah Pelajar")
plt.xlabel("Nama Pelajar")
plt.ylabel("Markah")

plt.show()
```
